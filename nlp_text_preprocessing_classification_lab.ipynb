```python
pd.set_option('display.max_colwidth', 100)
messages = pd.read_csv('C:/Users/USER/Desktop/spam.csv', encoding='latin-1')
messages = messages.drop(labels = ["Unnamed: 2","Unnamed: 3","Unnamed: 4"], axis=1)
messages.columns = ["label","text"]
messages.head()
```

```python
import string

def remove_punc(text):
    text = "".join([char for char in text if char not in string.punctuation])
    return text
```

```python
messages['clean_text'] = messages['text'].apply(lambda x: remove_punc(x))
messages.head()
```

```python
import re

def tokenize(text):
    tokens = re.split('\W+', text)
    return tokens

messages['tokenized_text'] = messages['clean_text'].apply(lambda x: tokenize(x.lower()))
messages.head()
```

```python
import nltk

stopwords = nltk.corpus.stopwords.words('english') #this loads the stopwords
```

```python
def remove_stopwords(tokenized_text):
    text = [word for word in tokenized_text if word not in stopwords]
    return text

messages['nostop_text'] = messages['tokenized_text'].apply(lambda x: remove_stopwords(x))
messages.head()
```
